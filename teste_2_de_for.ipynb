*Próximo desafio: Contador de números pares e ímpares
Agora vamos mudar o cenário. Quero que você tente construir um programa que percorra os números de 1 até 20 e conte quantos são pares e quantos são ímpares.*

```python
print("======== QUANTIDADE DE PARES E IMPARES ========\n")

numeros_pares = 0
numeros_impares = 0
qnt_par_impar = 0

for i in range(1, 21):
     if i % 2 == 0:
        numeros_pares += 1
     else:
        numeros_impares += 1


print("\n========= RESULTADO FINAL =========")
print(f"Números pares: {numeros_pares}")
print(f"Números Impares: {numeros_impares}")


```
