```python
print("========= FATURAMENTO DIÁRIO DA LOJA ==========")
print("===============================================")

valor_venda = 0

for i in range (1,7):

  faturamento = float(input(f"Digite o valor da venda {i}: "))

  valor_venda += faturamento

faturamento_dia = valor_venda / 6

print("========= RESULTADO DO FATURAMENTO ==========")
print("=============================================")
print(f"O faturamento no dia foi de R${valor_venda:.2f} reais")
print(f"E o ticket médio de venda hoje foi de R${faturamento_dia:.2f} reais")
```
