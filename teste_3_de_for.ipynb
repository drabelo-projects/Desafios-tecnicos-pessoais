Crie um programa que:
1. Peça ao usuário um número.
2. Utilize for para percorrer de 1 até 10.
3. Multiplique o número informado por cada valor do for.
4. Apresente a tabuada completa.

```python

print("======== Teste de Tabuada ========\n")

numero = int(input("Digite um número de 1 a 10: "))

for i in range(1, 11):
    print(f"\n{numero} x {i} = {numero * i}")
```
