```python
print("========= Conta Litros de Água =========\n")

total_litros = 0

for i in range (1,6):

 litros_d = float(input(f"Digite qual foi o consumo de água no dia {i}: "))

 total_litros += litros_d

mediaD = total_litros / 5


print("\n========= Conta Litros de Água =========\n")
print(f"O Consumo total de foi de {total_litros:.2f} Litros D´agua.")
print(f"E a média diária de consumo de foi de {mediaD:.2f} Litros D´agua. ")

```
