```python
print("============ CONTROLE DE VENDAS CAFETERIA =========\n")
print("===================================================\n")

faturamento_dia = 0

for i in range(1,8):

   valord = float(input(f"Informe qual foi o valor faturado no dia {i} R$ "))

   faturamento_dia += valord
   media = faturamento_dia / 7

print("\n============ RESULTADO DO FATURAMENTO DA CAFETERIA ============\n")
print("===============================================================")
print(f"O faturamento total foi de R${faturamento_dia:.2f} ")
print(f"A média diária de faturamento foi de R${media:.2f}")
```
