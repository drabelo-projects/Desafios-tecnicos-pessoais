Cenário:

Você está desenvolvendo um pequeno sistema para uma loja. O programa precisa registrar os preços de 5 produtos comprados por um cliente e, ao final, apresentar o valor total da compra.
Requisitos
1. Crie uma variável para armazenar o valor total da compra.
2. Utilize for para percorrer 5 produtos.
3. Solicite ao usuário o preço de cada produto.
4. Some os preços para obter o total.
5. Ao final, apresente o valor total da compra.

```python
print("======= REGISTRO DE PRODUTOS =======\n")

somap = 0

for i in range(1,6):

  precop = float(input(f"Digite o preço do produto {i}: "))

  somap += precop


print("\n1========= RESULTADO FINAL =========\n")
print(f"A soma de todos os produtos foi de R${somap:.2f} ")

```
