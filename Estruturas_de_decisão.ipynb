```python
x = 22
resto = x % 2

if resto == 0:
   print(f"O numero {x} é par")
else:
   print(f"O numero {x} é impar")
```

```python
nota = 5

if nota >= 9:
   print("Passou com maestria!")
elif nota >= 7:
   print("Passou Bem!")
elif nota >= 5:
   print("Passou por pouco!")
else:
    print("Já sabe, né? Não Passou!")
```

```python
lista_clientes = ['joao', 'maria', 'joana', 'ana']

cliente = 'ana'

if cliente in lista_clientes:
   print(f"O cliente {cliente} está na lista de clientes")
else:
   print(f"O cliente {cliente} não está na lista de clientes")
```

```python
cliente = 'Diogo'

if cliente in lista_clientes:
   print(f"O cliente {cliente} está na lista de clientes")
else:
   print(f"O cliente {cliente} não está na lista de clientes")
```

```python
conjunto = {1, 3, 5, 7}

elemento = 3

if elemento in conjunto:
   print(f"O elemento {elemento} está no conjunto")
else:
   print(f"O elemento {elemento} não está no conjunto")
```

```python
tupla = (1, 3, 5, 7)

if 10 in tupla:
    print("10 está na tupla")
else:
    print("10 não está na tupla")

```

```python
dicionario_clientes = {1: "joao", 2: "maria", 5: "Jose", 7: "ana"}

codigo_cliente = 5

if codigo_cliente in dicionario_clientes:
   print(f"O cliente de código {codigo_cliente} está no dicionário")
else:
   print(f"O cliente de código {codigo_cliente} não está no dicionário")
```
